```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Định nghĩa đường dẫn đến thư mục chứa các tệp CSV thuế quan đã làm sạch
tariff_csv_dir = r'E:\UIT\DS111\Raw data\tariff_csv'

# Liệt kê tất cả các tệp CSV trong thư mục
tariff_files = [f for f in os.listdir(tariff_csv_dir) if f.endswith('.csv')]

# Tạo một dictionary để lưu trữ các DataFrame
tariff_data = {}

print("Đang tải dữ liệu thuế quan...")
for file in tariff_files:
    file_path = os.path.join(tariff_csv_dir, file)
    df_name = file.replace('.csv', '') # Tên DataFrame sẽ là tên tệp không có phần mở rộng
    tariff_data[df_name] = pd.read_csv(file_path)
    print(f"Đã tải {file}")

print("Hoàn tất tải dữ liệu.")
```

```python
# Hiển thị thông tin cơ bản và 5 dòng đầu tiên của mỗi DataFrame
for name, df in tariff_data.items():
    print(f'\n--- Thông tin cho {name} ---')
    df.info()
    print(f'\n--- 5 dòng đầu tiên của {name} ---')
    print(df.head())
```

```python
# Hiển thị thống kê mô tả cho các cột số
for name, df in tariff_data.items():
    print(f'\n--- Thống kê mô tả cho {name} ---')
    print(df.describe())
```
